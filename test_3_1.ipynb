```python
formula = "(x0 | ~x1 | x3) & (~x0 | x2 | x5) & (x1 | x4 | ~x5) & (~x2 | x3 | x4) & (x0 | ~x4 | x5) & (~x1 | x2 | ~x3) & (x0 | x2 | ~x5) & (~x0 | x3 | x4) & (x1 | ~x2 | x5) & (~x1 | ~x4 | x3) & (x0 | ~x3 | x5) & (~x0 | x1 | ~x4) & (x2 | x3 | ~x4) & (~x2 | ~x3 | x5) & (x1 | x3 | x4) & (~x0 | ~x2 | x4) & (x0 | x1 | ~x5) & (~x1 | x2 | x4)"
```

```python
from itertools import product

clauses = formula.split(" & ")

def evaluate(bits):
    variables = {f"x{i}": bits[i] for i in range(6)}

    return all(
        any(
            (not variables[lit[1:]]) if lit.startswith("~")
            else variables[lit]
            for lit in clause.strip("()").split(" | ")
        )
        for clause in clauses
    )

solutions = [
    bits for bits in product([False, True], repeat=6)
    if evaluate(bits)
]

print("Formula valid:", all(len(c.strip("()").split(" | ")) == 3 for c in clauses))
print("Number of clauses:", len(clauses))
print("Satisfiable:", len(solutions) > 0)
print("Number of solutions:", len(solutions))
print("Solutions:", solutions)
```
